```python
'''Name: Amit Kumar, Regd No: 2341011129'''

from enum import Enum
import random

class Suit(Enum):
    HEARTS = "Hearts"
    DIAMONDS = "Diamonds"
    CLUBS = "Clubs"
    SPADES = "Spades"

class Card:
    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def get_value(self):
        if self.rank in ['Jack', 'Queen', 'King']:
            return 10
        elif self.rank == 'Ace':
            return 11
        else:
            return int(self.rank)

    def __repr__(self):
        return f"Card({self.suit.value}, {self.rank})"

class Deck:
    def __init__(self):
        self.cards = [Card(suit, rank) for suit in Suit for rank in ['Ace', '2', '3', '4', '5', '6', '7', '8', '9', '10', 'Jack', 'Queen', 'King']]

    def shuffle(self):
        random.shuffle(self.cards)

    def draw(self):
        if len(self.cards) > 0:
            return self.cards.pop()
        return None

class BlackjackGame:
    def __init__(self, num_players):
        self.deck = Deck()
        self.deck.shuffle()
        self.num_players = num_players
        self.player_hands = [[self.deck.draw(), self.deck.draw()] for _ in range(num_players)]
        self.dealer_hand = [self.deck.draw(), self.deck.draw()]

    def calculate_hand_value(self, hand):
        value = sum(card.get_value() for card in hand)
        aces = sum(1 for card in hand if card.rank == 'Ace')

        while value > 21 and aces > 0:
            value -= 10
            aces -= 1

        return value

    def player_hit(self, player_index):
        card = self.deck.draw()
        self.player_hands[player_index].append(card)
        print(f"Player {player_index + 1} hits and receives {card}.")
        if self.calculate_hand_value(self.player_hands[player_index]) > 21:
            return False
        return True

    def player_stand(self, player_index):
        pass

    def dealer_play(self):
        while self.calculate_hand_value(self.dealer_hand) <= 16:
            card = self.deck.draw()
            self.dealer_hand.append(card)
            print(f"Dealer hits and receives {card}.")

    def get_result(self):
        results = []
        dealer_total = self.calculate_hand_value(self.dealer_hand)

        for i in range(self.num_players):
            player_total = self.calculate_hand_value(self.player_hands[i])
            if player_total > 21:
                results.append(f"Player {i + 1}: Dealer wins.")
            elif dealer_total > 21:
                results.append(f"Player {i + 1}: Player wins.")
            elif player_total > dealer_total:
                results.append(f"Player {i + 1}: Player wins.")
            elif dealer_total > player_total:
                results.append(f"Player {i + 1}: Dealer wins.")
            else:
                results.append(f"Player {i + 1}: It's a tie.")

        return results

    def play_turn(self, player_index):
        print(f"\nPlayer {player_index + 1}'s Turn")
        print(f"Player {player_index + 1} hand: {self.player_hands[player_index]}")
        while True:
            total = self.calculate_hand_value(self.player_hands[player_index])

            if total > 21:
                break

            choice = input("Do you want to 'hit' or 'stand'? ").strip().lower()

            if choice == 'hit':
                self.player_hit(player_index)
                new_total = self.calculate_hand_value(self.player_hands[player_index])
                print(f"Updated hand: {self.player_hands[player_index]} (Total = {new_total})")
                if new_total > 21:
                    print("Bust!")
                    break
            elif choice == 'stand':
                self.player_stand(player_index)
                break
            else:
                print("Invalid choice. Please enter 'hit' or 'stand'.")

    def play_game(self):
        for i in range(self.num_players):
            print(f"Player {i + 1} hand: {self.player_hands[i]}")

        for i in range(self.num_players):
            self.play_turn(i)

        print("\nDealer's Turn")
        print(f"Dealer hand: {self.dealer_hand}")
        self.dealer_play()
        dealer_total = self.calculate_hand_value(self.dealer_hand)
        print(f"Dealer's final hand: {self.dealer_hand} (Total = {dealer_total})")

        print("\nResults")
        for res in self.get_result():
            print(res)
```
