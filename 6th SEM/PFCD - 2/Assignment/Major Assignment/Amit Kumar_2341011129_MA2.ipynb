```python
'''Name: Amit Kumar, Regd No: 2341011129'''
#a
def recursive_sum(n):
    if n <= 1: #base case
        return n
    else:
        return n+recursive_sum(n-1) #recursive step

def iterative_sum(n):
    total = 0
    while(n > 0):
        total += n
        n -= 1
    return total

n = 10
print(f"The sum of first {n} numbers using recursive sum is {recursive_sum(n)}")
print(f"The sum of first {n} numbers using iterative sum is {iterative_sum(n)}")

#b
'''Both recursive_sum(n) and iterative_sum(n) have a time complexity of: O(n) because both functions perform n operations to calculate the sum of the first n natural numbers.

In the recursive method, the function calls itself repeatedly until it reaches the base case, so the number of function calls increases linearly with n.
In the iterative method, the loop runs n times, adding each number to the total.'''

#c
'''Both recursion and iteration take O(n) time because both repeat the process n times to find the sum but the iterative method is more efficient because it uses less memory while in recursion, the function has to remember all the previous function calls until it reaches the base case.'''
```
