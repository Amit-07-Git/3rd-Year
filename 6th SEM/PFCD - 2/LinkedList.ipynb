```python
'''
1. Implement a Node class that contains 2 attributes data : to store data next : to store the reference to the next node

2. Using the Node class compose a Linked List class that contains the insert at beginning (), delete at beginning and traversal methods

3. Implement stack class using data classes. The stack class should contain a list attribute to store values. It also contains the push() pop() peek() and isEmpty methods

4. Using the stack class write a program to check if a given string of parentheses is valid or not.

Ex : '()' is valid, ')(' is invalid, '((' is invalid, ')' is invalid
'''
```

```python
#1
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

#2
class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        node1 = Node(data)
        node1.next = self.head
        self.head = node1
        
    def delete(self):
        if self.head is None:
            print("List is empty")
        self.head = self.head.next

    def traversal(self):
        if self.head is None:
            print("List is empty")
        current = self.head
        while current:
            print(current.data, end="->")
            current = current.next
        print("None")

lst = LinkedList()
lst.insert(55)
lst.insert(56)
lst.insert(57)

lst.traversal()

lst.delete()

lst.traversal()
```

```python
#3
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def push(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def pop(self):
        if self.is_empty():
            return "Stack is empty"
        popped_data = self.head.data
        self.head = self.head.next
        return popped_data

    def peek(self):
        if self.is_empty():
            return "Stack is empty"
        return self.head.data


stk = Stack()
stk.push(58)
stk.push(59)
stk.push(60)

print(f"Peek: {stk.peek()}")
print(f"Pop: {stk.pop()}")
print(f"New Top: {stk.peek()}")
```
