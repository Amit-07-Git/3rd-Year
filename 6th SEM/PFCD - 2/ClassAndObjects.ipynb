```python
from decimal import Decimal

class Account:

    savings = None
    no_of_acc = 0
    
    def __init__(self,name,balance):

        if balance < Decimal('0.00'):
            raise ValueError('Initial Balance must be >= 0.00')
        self.name = name
        self.balance = balance

        Account.no_of_acc += 1
        
    def deposit(self,amount):
        if (amount < Decimal('0.00')):
            raise ValueError('Amount must be non-negative')
        self.balance += amount
        
    def withdraw(self,amount):
        if (amount < Decimal('0.00')):
            raise ValueError('Withdraw amount must not be negative')
        if (amount > self.balance):
            raise ValueError('Withdraw amount must be <= Balance')
        self.balance -= amount

acct = Account("Batman",Decimal('999.00'))
acct = Account("Ironman",Decimal('456.00'))

Account.savings = Account("Spidermman Savings",Decimal('100'))

acct.deposit(Decimal('25.50'))
acct.withdraw(Decimal('40.00'))

print(acct.balance)
print("Savings:",Account.savings.balance)
print("Number of accounts: ", Account.no_of_acc)
```
