```python
key = input("Enter The Key: ").upper()
text = input("Enter The Text: ").upper()

s = ""
for c in key + "ABCDEFGHIKLMNOPQRSTUVWXYZ":
    if c == "J": c = "I"
    if c not in s and c.isalpha():
        s += c

m = [list(s[i:i+5]) for i in range(0, 25, 5)]

print("\nKey Matrix:")
for row in m:
    formatted_row = [(ch + "/J" if ch == 'I' else ch).ljust(3) for ch in row]
    print(" ".join(formatted_row))

t = text.replace("J", "I").replace(" ", "")
pairs = []
i = 0
while i < len(t):
    a = t[i]
    if i + 1 < len(t):
        b = t[i+1]
        if a == b:
            pairs.append(a + 'X')
            i += 1
        else:
            pairs.append(a + b)
            i += 2
    else:
        pairs.append(a + 'X') 
        i += 1

print("\nPlaintext pairs:")
print(" ".join(pairs))

def pos(x):
    for i in range(5):
        for j in range(5):
            if m[i][j] == x:
                return i, j


enc = ""
for a, b in pairs:
    r1, c1 = pos(a)
    r2, c2 = pos(b)
    if r1 == r2:
        enc += m[r1][(c1+1)%5] + m[r2][(c2+1)%5]
    elif c1 == c2:
        enc += m[(r1+1)%5][c1] + m[(r2+1)%5][c2]
    else:
        enc += m[r1][c2] + m[r2][c1]

print("\nEncrypted:", enc)

dec = ""
for i in range(0, len(enc), 2):
    a, b = enc[i], enc[i+1]
    r1, c1 = pos(a)
    r2, c2 = pos(b)
    if r1 == r2:
        dec += m[r1][(c1-1)%5] + m[r2][(c2-1)%5]
    elif c1 == c2:
        dec += m[(r1-1)%5][c1] + m[(r2-1)%5][c2]
    else:
        dec += m[r1][c2] + m[r2][c1]

final = ""
k = 0
while k < len(dec):
    if k < len(dec)-2 and dec[k] == dec[k+2] and dec[k+1] == 'X':
        final += dec[k]
        k += 2
    else:
        final += dec[k]
        k += 1
if final.endswith('X'): final = final[:-1]

print("Decrypted:", final)
```
