```python
#Q9
def bin_dec(binary_no):
    return int(binary_no,2)
def dec_bin(decimal_no):
    if decimal_no == 0:
        return "0"
    binary = ""
    while decimal_no != 0:
        rem = decimal_no % 2
        binary = str(rem) + binary
        decimal_no //= 2
    return binary
binary_no = input("Enter a binary number: ")
print("Decimal:", bin_dec(binary_no))
decimal_no = int(input("Enter a decimal number: "))
print("Binary:", dec_bin(decimal_no))
```

```python
#Q10
def permute(s, answer=""):
    if len(s) == 0:
        print(answer,end=" ")
        return
    for i in range(len(s)):
        ch = s[i]
        left = s[:i]
        right = s[i+1:]
        rest = left + right
        permute(rest, answer + ch)
permute("abc")
```

```python
#Q13
def roman_to_int(s):
    values = {'I': 1, 'V': 5, 'X': 10, 'L': 50,
              'C': 100, 'D': 500, 'M': 1000}
    total = 0
    prev_value = 0
    for ch in reversed(s):
        value = values[ch]
        if value < prev_value:
            total -= value
        else:
            total += value
        prev_value = value
    return total
roman = input("Enter a Roman numeral: ").upper()
print("Integer value:", roman_to_int(roman))
```

```python
#Q14
def is_armstrong(num):
    return num == sum(int(d)**len(str(num)) for d in str(num))
n = int(input("Enter a number: "))
if is_armstrong(n):
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
#Q15
def fibonacci(n):
    a, b = 0, 1
    for i in range(n):
        a, b = b, a + b
    return a
n = int(input("Enter the position (n): "))
print("The", n, "th Fibonacci number is:", fibonacci(n))
```

```python
#Q16
def calculator(a, b, op):
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        if b != 0:
            return a / b
        else:
            return "Cannot divide by zero!"
    else:
        return "Invalid operator!"
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
op = input("Enter operator (+, -, *, /): ")
print("Result:", calculator(a, b, op))
```

```python
#Q18
def is_anagram(str1, str2):
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()
    return sorted(str1) == sorted(str2)
word1 = input("Enter first string: ")
word2 = input("Enter second string: ")
if is_anagram(word1, word2):
    print("They are anagrams!")
else:
    print("They are not anagrams.")
```

```python
#Q19
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
print("GCD:", gcd(num1, num2))
```

```python
#Q20
def print_primes():
    for num in range(2, 101):
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num, end=" ")

print_primes()
```

```python
#Q24
def remove_punctuation(s):
    punctuation = '''!()-[]{};:'"\\,<>./?@#$%^&*_~'''
    result = ""
    for char in s:
        if char not in punctuation:
            result += char
    return result
text = "Hello, world! How'z the josh?"
clean_text = remove_punctuation(text)
print(clean_text)
```

```python
#Q25
def are_coprime(a, b):
    gcd = 1
    for i in range(2, min(a, b) + 1):
        if a % i == 0 and b % i == 0:
            gcd = i
            break
    return gcd == 1
if are_coprime(14,15):
    print("Coprime")
else:
    print("Not Coprime")
```

```python
#Q26
def replace_vowels(s):
    vowels = "aeiouAEIOU"
    result = ""
    for char in s:
        if char in vowels:
            result += "*"
        else:
            result += char
    return result
print(replace_vowels("Hello World"))
```

```python
#Q27
def digits_to_words(num):
    words = ["zero", "one", "two", "three", "four",
             "five", "six", "seven", "eight", "nine"]
    num_str = str(num)
    result = ""
    for digit in num_str:
        result += words[int(digit)] + " "
    return result
print(digits_to_words(85))
print(digits_to_words(123))
```

```python
#Q28
def shift_letters(s):
    result = ""
    for char in s:
        if char == 'z':
            result += 'a'
        else:
            result += chr(ord(char) + 1)
    return result
print(shift_letters("python"))
```

```python
#Q29
def is_perfect(n):
    sum_divisors = 0
    for i in range(1, n):
        if n % i == 0:
            sum_divisors += i
    return sum_divisors == n
num = 28
if is_perfect(num):
    print(num, "is a perfect number")
else:
    print(num, "is not a perfect number")
```

```python
#Q30
def product_of_digits(n):
    product = 1
    while n > 0:
        digit = n % 10
        product *= digit
        n = n // 10
    return product
num = 1234
print("Product of digits of", num, "is", product_of_digits(num))
```
