```python
#Q6
num = int(input("Enter an integer: "))
if num <= 1:
    print(num,"is not a prime number")
elif num == 2:
    print(num,"is a prime number")
else:
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num,"is a prime number")
    else:
        print(num,"is not a prime number")
```

```python
#Q7
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
def sum_prime(n):
    total = 0
    for num in range(2, n):
        if is_prime(num):
            total += num
    return total
n = int(input("Enter a number: "))
result = sum(n)
print("The sum of all prime numbers below",n,"is",result)
```

```python
#Q9
def remainder():
    try:
        n = int(input("Enter a number: "))
        match n % 5:
            case r:
                if 0 <= r <= 4:
                    print("The remainder is",r)
    except ValueError:
        print("Invalid input")
remainder()
```

```python
#Q10
def substrings(s):
    n = len(s)
    for i in range(n):
        for j in range(i + 1, n + 1):
            print(s[i:j], end=" ")
string = input("Enter a string: ")
print("All substrings are:")
substrings(string)
```

```python
#Q11
a = int(input("Enter first Number:"))
b = int(input("Enter second Number:"))
op = input("Enter an operator(+,-,*,/):")
match op:
    case '+': print("The Addition is",a + b)
    case '-': print("The Substraction is",a - b)
    case '*': print("The Multiplication is",a * b)
    case '/':
        if b!=0:
            print("The Division is",a / b)
        else:
            print("Error: Division by zero")
    case _: print("Invalid operator")
```

```python
#Q14
room = input("Enter room type(Standard/Deluxe/Suite): ").lower()
nights = int(input("Enter number of nights: "))
season = input("Enter season(peak/normal/off): ").lower()
loyalty = input("Loyalty Member?(yes/no): ").lower()
if room == "standard":
    cost = 100
elif room == "deluxe":
    cost = 150
elif room == "suite":
    cost = 250
else:
    print("Invalid Room Type")
total = cost*nights
if nights>7:
    total *= 0.8
elif nights>3:
    total *= 0.9
if season == "peak":
    total *= 1.2
elif season =="off":
    total *= 0.85
if loyalty == "yes":
    total *= 0.95
print("Final cost: $",total)
```

```python
#Q15
num = int(input("Enter a Number: "))
sum = 0
for i in range (1,num):
    if num%i == 0:
        sum += i
if sum == num:
    print(num,"is a perfect number")
else:
    print(num,"is not a perfect number")
```

```python
#Q16
#(a)
import math
x = float(input("Enter the value of x: "))
n = int(input("Enter number of terms: "))
sum_a = 1
sign = -1
for i in range(1, n + 1):
    term = (x ** (2 * i)) / math.factorial(2 * i)
    sum_a = sum_a + (sign * term)
    sign = -sign
print("Sum of series (a) =", sum_a)

#(b)
import math
x = float(input("Enter the value of x: "))
n = int(input("Enter number of terms: "))
sum_b = 1
for i in range(1, n + 1):
    term = (x ** i) / math.factorial(i)
    sum_b = sum_b + term
print("Sum of series (b) =", sum_b)

#c
n = int(input("Enter number of terms: "))
sum_c = 0
num = 1
sign = 1
for i in range(n):
    sum_c = sum_c + (sign * num)
    num = num + 2
    sign = -sign
print("Sum of series (c) =", sum_c)
```

```python
#Q21
n = int(input("Enter a number: "))
fact,i = 1,1
while fact < n:
    i += 1
    fact *= i
if fact == n:
    print(n, "is a factorial number (",i,"!)")
else:
    print(n, "is not a factorial number.")
```

```python
#Q22
n = int(input("Enter a number: "))
while n >= 10:
    s = 0
    while n > 0:
        s += n % 10
        n //= 10
    n = s
print("Single-digit sum:", n)
```

```python
#Q23
amount = int(input("Enter amount: "))
if amount % 10 != 0:
    print("Amount must be multiple of 10")
elif amount > 1000:
    print("Insufficient funds in ATM")
else:
    for note in [100, 50, 20, 10]:
        count = amount // note
        if count:
            print(count,"->",note,"Rs note")
        amount %= note
```
