F-String Presentation Types

1 : Python supports precision only for floating-point and Decimal values.
2 : Formatting is type dependent—if you try to use .2f to format a string like 'hello', a ValueError occurs.

Multiple Presentation types exist. We will focus on Integers, Characters, Strings, Floating and Decimal presentation types.

The d presentation type formats integer values as strings.
There also are integer presentation types (b, o and x or X) that format integers using the binary, octal or hexadecimal number systems.
The s presentation type is the default. If you specify s explicitly, the value to format must be a variable that references a string, an expression that produces a string or a string literal, as in the first placeholder {} below. 
If you do not specify a presentation type, as in the second placeholder below, non-string values like the integer 7 are converted to strings

```python
print(f'{10:d}') # represents 10 in decimal
print(f'{10:b}') # represents 10 in binary
print(f'{10:o}') # represents 10 in octal
print(f'{10:x}') # represents 10 in hexadecimal
print(f'{65:c} {97:c}') # represents 65 and 97 in ASCII
print(f"{'hello':s}") # represents a string
```

The 'f' presentation type is used to control the number of digits after the decimal point. 
There is another presentation type 'e' which is used to represent scientific/ exponential notation

```python
print(f'{Decimal('10247667789990.02'):.4f}') # Prints 10245.0200 
print(f'{Decimal('10247667789990.02'):.3e}') # Prints 1.025e+13
print(f'{Decimal('10247667789990.02'):.4e}') # Prints 1.0248e+13
```

Field Widths and alignment
-> Python right aligns the numbers and left-aligns the strings by default.

```python
print(f'{2:5d}')
print(f'{24:5d}') # 5d says represent 24 as a 5 digit number and hide the zeros that are left-most, rightmost zeros are visible
print(f'{240:5d}')
print(f'{2400:5d}')
print(f'{24000:5d}')
print(f'{240000:5d}') # this will not be aligned. Value accuracy is more important than presentation
```

-> Alignment can be made explicitly with the <, > and ^ symbols:

```python
print(f'{15:<4d}') 
print(f'{150:<4d}')
print(f'{1500:<4d}')
print(f'{1500:<5d}\n')

print(f'{15:^7d}') 
print(f'{150:^7d}')
print(f'{1500:^7d}')
print(f'{15000:^7d}\n')

print(f'{'JonathanByers':^017s}')
print(f'{'WillByers':^017s}')
```

Numeric Formatting
+ and be appended to the numbers before printing.
, can be used as thousands separator

```python
print(f'{24:+10d}')
print(f'{24:+010d}')
print(f'{123456789:,d}')
print(f'{123456789:018,d}')
print(f'{789123.2334456:,.3f}')
print(f'{789123.2334456:015,.3f}')
```

String's Format Method

-> F strings were added to python in 3.6.
-> Before that formatting was performed using the string's .format()
-> A format string may contain multiple placeholders, in which case the format method’s arguments correspond to the placeholders from left to right
-> The format string can reference specific arguments by their position in the format method’s argument list, starting with position 0
-> You can reference keyword arguments by their keys in the placeholders

```python
print("{:.2f}".format(13.33456)) # prints(13.33)
print("{} is {}".format('My name is', 'Mike Flannagan'))
print('{0} {0} {1}'.format('Happy', 'Borkday'))
print('{a} {b}'.format(a="Belgian", b="Mallinois"))
```

Substring Methods
1 : Counting Occurences
-> .count() returns the number of times the argument occurs in the string on which the method is called.

2 : Locating substring within a string
-> .index() searches for a substring within a string and returns the first index where it is found.
-> .rindex() searches for a substring within a string but starting from the end.
-> if the substring is not found a value error occurs.
-> it is better to use the .find() and .rfind() method for locating substrings, as they do not return value error if the substring doesn't exist.

3 : Determining whether a string contains a substring
-> the in operator can be used to check if a substring exist within a string
-> not in also can be used
-> returns a boolean value

4 : Locating a substring at the beginning or end of a string
-> .startswith() and .endswith() return True if the string starts with or ends with a specified substring 

```python
text = 'to be or not t o be that is the question, to be happy is one thing, to be sad is yet another'
print(text.count('to'))             # count the frequency of the word to starting from 0 till the end
print(text.count('to', 15))         # count the frequency of the word to starting from index 15 upto end
print(text.count('to', 0, 10))      # count the frequency of the word to starting from 0 upto 10 but not including 10
print(text.index('is'))
print(text.index('be', 15))
print(text.index('or', 0, 10))
# print(text.index('yetty'))         # will cause an error
print(text.find('yetty'))           # returns -1 if substring doesn't exist
print(text.find('is'))
print(text.rfind('is'))
print('sad' in text)
print('HAPPY' not in text)
print(text.startswith('TO'))
print(text.endswith('another'))
```

Replacing Substrings

-> we cannot directly modify a string in memory. 
-> s = 'abc' ; s[0] = 'A' will not work as strings are immutable in python. 
-> so to modify strings, we need to use the methods associated with the string object or the string class to modify strings
-> the .replace() method returns new string replacing the all the occurences of the old substring with a new substring
-> it takes two arguments, the first argument describes the substring to be replaced, the second argument is the value that is to be written over the first substring.

```python
v = '1\t2\t3\t4\t5'
print(v.replace('\t', ' <-x-x-> ')) # replace \t with <-x-x->
v
```

-> the default delimiter is the ' ' 
-> the second argument in the .split() controls the number of splits. Once the value is reached it doesn't break the remaining part of string till the end.
-> the .rsplit() works from end to the beginnning.

Partitioning Strings

-> the .partition() and rpartition() splits the string into a tuple of three substrings based on the method's separator argument.
-> the partitions are
    1 : the LHS of the separator (but not including the separator) passed to the method
    2 : the separator
    3 : the RHS of the separator (but not including the separator) upto the end

String Method .splitlines()

-> A docstring is a string that covers multiple lines.
-> To break apart a docstring or single/double quoted strings containing multiple newlines the .splitlines() can be used.
-> It breaks the entire line into a list of lines.  
-> Passing True to .splitlines() keeps the newlines at the end of each string

```python
t = 'a for apple, b for ball, c for cat'
print(t.split())
print(t.split(','))
print(t.split(' for '))
print(t.split('for', 2))
```

```python
fruits = ['apple', 'cherry', 'avocado', 'kiwi']

print(''.join(fruits))            #all the strings will merged together without any spaces
print(' spiderman '.join(fruits)) #all the strings will merged together with the string spiderman in between the names of fruits
print(' x-> '.join(fruits))       #all the strings will merged together with the string x-> in between the names of fruits
```

```python
st = 'George Marshall Cooper Jr. : 56, 52, 70'
print(st.partition(':'))
print()
for item in st.partition(','):
    print(item)
```
